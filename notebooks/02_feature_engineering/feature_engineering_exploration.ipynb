## 验证文本信息

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import zlib  # 引入压缩库
from matplotlib.dates import DateFormatter

def calculate_compression_ratio(text_series):
    """
    计算 '文本压缩比' (Compression Ratio)
    原理：利用 zlib (LZ77) 算法检测重复模式。
    不管你加什么表情、标点，只要核心“长字符串”重复，压缩比就会骤降。
    
    返回：压缩后大小 / 原始大小
    范围：0.0 (极度重复) ~ 1.0 (完全随机)
    预期：正常对话约 0.4~0.6；刷屏会掉到 0.1 以下。
    """
    # 1. 简单拼接
    # 不需要复杂的清洗，直接把所有评论拼成一个巨大的字符串
    # 甚至保留标点更有利于检测，因为刷屏往往连标点都复制
    full_text = "".join([str(t) for t in text_series])
    
    if not full_text: return np.nan
    
    # 转换为字节
    original_bytes = full_text.encode('utf-8')
    original_size = len(original_bytes)
    
    if original_size < 100: # 字节数太少，压缩没意义
        return np.nan
        
    # 2. 核心：进行压缩
    compressed_bytes = zlib.compress(original_bytes)
    compressed_size = len(compressed_bytes)
    
    # 3. 计算比率
    ratio = compressed_size / original_size
    return ratio

def verify_raw_data_compression(df, start_date, end_date):
    print(f"🩺 正在进行 V3 版体检：基于 LZ77 压缩算法检测视觉冲击...")
    
    # 数据切片
    df['dt'] = pd.to_datetime(df['timestamp'], errors='coerce')
    mask = (df['dt'] >= start_date) & (df['dt'] <= end_date)
    target_df = df.loc[mask].copy()
    target_df.set_index('dt', inplace=True)
    
    if target_df.empty:
        print("❌ 无数据")
        return

    # 聚合：建议 15分钟 或 30分钟
    RESAMPLE_FREQ = '15T'
    
    # 计算流量
    vol_series = target_df['comment_content'].resample(RESAMPLE_FREQ).count()
    
    # 计算压缩比
    comp_series = target_df['comment_content'].resample(RESAMPLE_FREQ).apply(calculate_compression_ratio)

    # --- 绘图 ---
    sns.set(style="whitegrid", font='SimHei')
    fig, ax1 = plt.subplots(figsize=(14, 6))

    # 左轴：流量
    ax1.set_xlabel('时间 (Time)')
    ax1.set_ylabel('评论量 (Volume)', color='dimgrey')
    ax1.bar(vol_series.index, vol_series.values, width=0.008, color='silver', label='评论量', alpha=0.5)
    
    # 右轴：压缩比
    ax2 = ax1.twinx()
    # 颜色选深红，代表警报
    color_comp = '#d62728'
    ax2.set_ylabel('信息密度 (Compression Ratio)\n越低代表重复度越高', color=color_comp, fontsize=12)
    
    # 画折线
    ax2.plot(comp_series.index, comp_series.values, 
             color=color_comp, marker='o', linestyle='-', linewidth=2.5, label='文本压缩比 (LZ77)')
    
    # 画正常基准线 (通常正常人类对话在 0.4 左右)
    ax2.axhline(y=0.4, color='green', linestyle='--', alpha=0.5, label='正常对话基准 (0.4)')
    # 画刷屏警戒线
    ax2.axhline(y=0.15, color='orange', linestyle='--', alpha=0.8, label='刷屏警戒线 (0.15)')
    
    ax2.tick_params(axis='y', labelcolor=color_comp)
    # y轴范围不需要固定，让它自适应展示差异
    # ax2.set_ylim(0, 1.0) 

    # 格式化
    date_format = DateFormatter("%m-%d %H:%M")
    ax1.xaxis.set_major_formatter(date_format)
    fig.autofmt_xdate()

    plt.title(f"验证3 (终极版): 流量 vs 文本压缩比 (感知视觉重复)", fontsize=14)
    plt.tight_layout()
    plt.show()
    
    # 打印最严重的时刻
    min_ratio = comp_series.min()
    print(f"\n🚨 监测报告:")
    print(f"   - 最低信息密度: {min_ratio:.4f}")
    print(f"   - 解释: 只有 {min_ratio*100:.1f}% 的字节是新信息，其余全是重复废话。")
    print(f"   - 发生时刻: {comp_series.idxmin()}")

# 调用
# verify_raw_data_compression(df, '2025-11-01 18:00', '2025-11-02 06:00')
```

## 验证图像重复度

```python
def calculate_visual_repetition(image_url_series):
    """
    计算 '图片重复度'
    输入：包含图片URL列表的 Series
    """
    all_urls = []
    
    # 1. 展开 URL (假设数据里是 list 或逗号分隔字符串)
    for item in image_url_series:
        if not item: continue
        # 如果是字符串 list 格式 "['url1', 'url2']" 需要 eval 或 json.loads
        # 这里假设已经是 list 或者以逗号分隔的字符串
        if isinstance(item, str):
            urls = item.split(',') # 根据实际分隔符调整
        elif isinstance(item, list):
            urls = item
        else:
            continue
        all_urls.extend([u.strip() for u in urls])
            
    if not all_urls: return np.nan # 该时间段没图
    
    total_imgs = len(all_urls)
    unique_imgs = len(set(all_urls))
    
    # 2. 计算同质性指标
    # 1.0 代表所有图都不一样 (Cluster Density Low)
    # 0.0 代表所有图都一样 (Cluster Density High)
    # 为了符合“异常越高”的直觉，我们可以算 Repetition Rate
    repetition_rate = 1.0 - (unique_imgs / total_imgs)
    
    return repetition_rate

def verify_visual_impact(df, start_date, end_date):
    print(f"🩺 正在进行 V5 版体检：视觉模因同质性检测 (Visual Homogeneity)...")
    
    df['dt'] = pd.to_datetime(df['timestamp'], errors='coerce')
    mask = (df['dt'] >= start_date) & (df['dt'] <= end_date)
    target_df = df.loc[mask].copy()
    target_df.set_index('dt', inplace=True)
    
    RESAMPLE_FREQ = '15T'
    
    # 只看有图的微博/评论
    # 假设列名是 image_urls
    
    # 计算含图率 (Visual Density)
    def has_image(x):
        if pd.isna(x) or x == '' or x == '[]': return 0
        return 1
    
    # 含图贴数量
    img_vol_series = target_df['image_urls'].resample(RESAMPLE_FREQ).apply(lambda x: x.apply(has_image).sum())
    
    # 图片重复率 (Visual Repetition)
    visual_rep_series = target_df['image_urls'].resample(RESAMPLE_FREQ).apply(calculate_visual_repetition)

    # --- 绘图 ---
    fig, ax1 = plt.subplots(figsize=(14, 6))
    
    # 柱状图：含图贴数量
    ax1.bar(img_vol_series.index, img_vol_series.values, width=0.008, color='purple', alpha=0.3, label='含图评论数')
    ax1.set_ylabel('含图样本量', color='purple')
    
    # 线图：图片重复率
    ax2 = ax1.twinx()
    color_vis = '#d62728'
    ax2.plot(visual_rep_series.index, visual_rep_series.values, 
             color=color_vis, marker='s', linestyle='-', label='图片URL重复率')
    
    # 阈值线
    ax2.axhline(y=0.3, color='orange', linestyle='--', label='警告线 (30%重复)')
    
    ax2.set_ylabel('视觉同质性 (0~1)', color=color_vis)
    ax2.set_ylim(0, 1.1)
    
    plt.title(f"验证5: 视觉炸弹 — 大家在发同一张图吗？", fontsize=14)
    plt.show()
```

## 验证符号

```python
import re
import pandas as pd
import numpy as np

def calculate_weighted_intensity(text_series):
    """
    V7 升级版：情绪烈度加权计算
    逻辑：
    1. 统计每条评论中敏感符号的个数。
    2. 设置“饱和阈值” (CAP = 5)。超过5个按5个算，防止单个疯子刷屏破坏数据。
    3. 返回该时间段内的总烈度分。
    """
    if text_series.empty: return 0
    
    # 转换为字符串
    text_series = text_series.astype(str)
    
    # 定义符号正则
    pattern = r'[❗!！\?？🕯️]'
    
    # 内部函数：计算单条文本的得分
    def get_score(text):
        # 找出所有匹配的符号
        matches = re.findall(pattern, text)
        count = len(matches)
        # 截断逻辑：最大只给 5 分 (防止一个评论带100个感叹号)
        return min(count, 5)
    
    # 应用到 Series 并求和
    total_intensity = text_series.apply(get_score).sum()
    
    return total_intensity
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.dates import DateFormatter
from utilis_preprocess import conbine_post_comment


# --- 核心算法：基尼系数计算 ---
def calculate_gini(user_series):
    """
    计算用户发言的基尼系数 (Gini Coefficient)
    输入: user_id 的 Series (在该时间窗口内的所有发言用户ID)
    输出: 0.0 ~ 1.0
    逻辑:
      - 如果没人发言，返回 NaN
      - 如果只有1个人发言，Gini = 0 (完全平等? 其实是完全垄断，但公式在n=1时通常归0，这里我们特殊处理)
        * 修正：对于异常检测，如果只有1个人发了100条，我们希望 Gini 很高。
        * 但标准 Gini 是衡量分布不平等的。
        * 让我们用标准公式，结合流量看即可。
    """
    if user_series.empty: return np.nan
    
    # 统计每个用户的发言次数
    # 例如: User A 发了 10 条, User B 发了 1 条 -> counts = [10, 1]
    counts = user_series.value_counts().values
    
    n = len(counts)
    if n == 0: return np.nan
    if n == 1: return 0.0 # 只有一个用户，不存在“不平等”（或者说无法计算分布差异）
    
    # 排序 (从小到大)
    counts = np.sort(counts)
    
    # Gini 计算公式
    # G = (2 * sum(i * y_i) / (n * sum(y_i))) - (n + 1) / n
    # 这里的 i 是 1 到 n
    index = np.arange(1, n + 1)
    gini = (2 * np.sum(index * counts)) / (n * np.sum(counts)) - (n + 1) / n
    
    return gini

# --- 验证主程序 ---
def verify_participation_concentration(df1,df2, start_date, end_date):
    print(f"🔬 正在验证特征：参与度集中性 (基尼系数)...")
    
    df,df_post,df_comment = conbine_post_comment(df1,df2)
    # return df
    # 1. 数据预处理
    df['dt'] = pd.to_datetime(df['timestamp'], errors='coerce')
    mask = (df['dt'] >= start_date) & (df['dt'] <= end_date)
    target_df = df.loc[mask].copy()
    target_df.set_index('dt', inplace=True)
    
    if target_df.empty:
        print("❌ 该时间段无数据")
        return

    # 设定时间窗口 (建议 30分钟，太短会导致样本太少 Gini 失效)
    RESAMPLE_FREQ = '15T'
    

    
    # 2. 计算指标
    # A. 流量 (Total Volume) - 作为背景参考
    # B. 参与人数 (Unique Users) - 辅助参考
    # C. 基尼系数 (Gini) - 核心特征
    # vol_series = target_df['user_id'].resample(RESAMPLE_FREQ).count()
    # unique_user_series = target_df['user_id'].resample(RESAMPLE_FREQ).nunique()
    # gini_series = target_df['user_id'].resample(RESAMPLE_FREQ).apply(calculate_gini)
    

    vol_series = target_df['author_id'].resample(RESAMPLE_FREQ).count()
    unique_user_series = target_df['author_id'].resample(RESAMPLE_FREQ).nunique()
    gini_series = target_df['author_id'].resample(RESAMPLE_FREQ).apply(calculate_gini)

    # --- 绘图 ---
    fig, ax1 = plt.subplots(figsize=(14, 6))

    # 背景：流量柱状图
    ax1.bar(vol_series.index, vol_series.values, width=0.015, color='gainsboro', label='总评论量 (Volume)', alpha=0.8)
    ax1.set_ylabel('评论量 / 发言数', color='gray')
    ax1.set_xlabel('时间')

    # 右轴：基尼系数
    ax2 = ax1.twinx()
    color_gini = '#9467bd' # 紫色
    
    # 画 Gini 曲线
    ax2.plot(gini_series.index, gini_series.values, 
             color=color_gini, marker='o', linestyle='-', linewidth=2, label='基尼系数 (集中度)')
    
    # 辅助线：0.5 通常被认为是贫富差距/集中度的警戒线
    ax2.axhline(y=0.5, color=color_gini, linestyle=':', alpha=0.5, label='集中度警戒线 (0.5)')
    
    ax2.set_ylabel('基尼系数 (0=平均, 1=垄断)', color=color_gini, fontsize=12)
    ax2.set_ylim(0, 1.1)
    
    # 格式化
    ax1.xaxis.set_major_formatter(DateFormatter("%m-%d %H:%M"))
    
    # 合并图例
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    
    plt.title(f"验证: 参与度集中性 — 是'水军刷屏'还是'全员暴动'?", fontsize=14)
    plt.tight_layout()
    plt.show()

    # --- 打印分析 ---
    max_gini_time = gini_series.idxmax()
    max_gini_val = gini_series.max()
    print(f"\n📊 分析报告:")
    print(f"   - 集中度最高时刻: {max_gini_time} (Gini: {max_gini_val:.2f})")
    if pd.notna(max_gini_time):
        vol_at_peak = vol_series.loc[max_gini_time]
        users_at_peak = unique_user_series.loc[max_gini_time]
        print(f"   - 当时情况: {vol_at_peak} 条评论来自 {users_at_peak} 个用户")
        print(f"   - 平均每人发言: {vol_at_peak/users_at_peak:.1f} 条")
```

## 验证长度离散程度

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.dates import DateFormatter

# --- 核心算法：长度标准差 ---
def calculate_length_std(text_series):
    """
    计算文本长度的标准差 (Standard Deviation)
    输入: 评论内容 Series
    输出: 长度的标准差 (数值越大，长短差异越明显；数值越小，越像复制粘贴)
    """
    if text_series.empty: return np.nan
    
    # 1. 计算每条评论的长度
    # 转换为字符串，处理空值
    lengths = text_series.astype(str).apply(len)
    
    # 2. 如果样本太少（比如只有1-2条），标准差没有意义，返回 NaN
    if len(lengths) < 3: return np.nan
    
    # 3. 计算标准差
    return lengths.std()

# --- 验证主程序 ---
def verify_length_variance(df, start_date, end_date):
    print(f"📏 正在验证特征：文本长度离散度 (Length Standard Deviation)...")
    
    # 1. 数据预处理
    df['dt'] = pd.to_datetime(df['timestamp'], errors='coerce')
    mask = (df['dt'] >= start_date) & (df['dt'] <= end_date)
    target_df = df.loc[mask].copy()
    target_df.set_index('dt', inplace=True)
    
    if target_df.empty:
        print("❌ 该时间段无数据")
        return

    # 设定时间窗口 (建议 30分钟)
    RESAMPLE_FREQ = '15T'
    
    # 2. 计算指标
    # A. 流量 (Volume)
    vol_series = target_df['comment_content'].resample(RESAMPLE_FREQ).count()
    
    # B. 长度标准差 (Length Std) - 核心特征
    len_std_series = target_df['comment_content'].resample(RESAMPLE_FREQ).apply(calculate_length_std)

    # --- 绘图 ---
    fig, ax1 = plt.subplots(figsize=(14, 6))

    # 背景：流量柱状图
    ax1.bar(vol_series.index, vol_series.values, width=0.015, color='gainsboro', label='总评论量 (Volume)', alpha=0.8)
    ax1.set_ylabel('评论量', color='gray')
    ax1.set_xlabel('时间')

    # 右轴：长度标准差
    ax2 = ax1.twinx()
    color_std = '#2ca02c' # 绿色
    
    # 画标准差曲线
    ax2.plot(len_std_series.index, len_std_series.values, 
             color=color_std, marker='^', linestyle='-', linewidth=2, label='长度标准差 (离散度)')
    
    # 辅助线：警戒线 (经验值：如果标准差低于 5，说明大家说话长度几乎一样)
    ax2.axhline(y=5, color='red', linestyle='--', alpha=0.5, label='僵尸/刷屏警戒线 (<5)')
    
    ax2.set_ylabel('长度标准差 (字数)', color=color_std, fontsize=12)
    
    # 格式化
    ax1.xaxis.set_major_formatter(DateFormatter("%m-%d %H:%M"))
    
    # 合并图例
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    
    plt.title(f"验证: 文本形态特征 — 大家是在'讨论'(高离散) 还是在'复读'(低离散)?", fontsize=14)
    plt.tight_layout()
    plt.show()

    # --- 打印极值分析 ---
    min_std_time = len_std_series.idxmin()
    min_std_val = len_std_series.min()
    
    print(f"\n📊 分析报告:")
    print(f"   - 长度最整齐划一的时刻: {min_std_time}")
    print(f"   - 最低标准差: {min_std_val:.2f} (意味着大部分评论长度差异极小)")
    
    if pd.notna(min_std_time):
        vol_at_peak = vol_series.loc[min_std_time]
        print(f"   - 当时流量: {vol_at_peak} 条")
        if vol_at_peak > 50 and min_std_val < 5:
            print("   - 🚨 判定: 高度疑似有组织刷屏！")
        else:
            print("   - ✅ 判定: 正常或样本过少")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.dates import DateFormatter

# --- 核心分类函数 ---
def categorize_length(text):
    """
    将文本长度分类：
    0: 短文本 (Short) < 10字 (宣泄/跟风)
    1: 中文本 (Medium) 10-50字 (普通讨论)
    2: 长文本 (Long) > 50字 (小作文/深度维权)
    """
    if pd.isna(text): return None
    l = len(str(text))
    if l < 10: return 'Short'
    elif l > 50: return 'Long'
    else: return 'Medium'

# --- 验证主程序 ---
def verify_text_length_layers(df, start_date, end_date):
    print(f"📚 正在进行 V10 版体检：长短文本分层检测 (小作文 vs 口吐芬芳)...")
    
    # 1. 数据预处理
    df['dt'] = pd.to_datetime(df['timestamp'], errors='coerce')
    mask = (df['dt'] >= start_date) & (df['dt'] <= end_date)
    target_df = df.loc[mask].copy()
    target_df.set_index('dt', inplace=True)
    
    if target_df.empty:
        print("❌ 无数据")
        return

    RESAMPLE_FREQ = '15T'
    
    # 2. 先给每一行打标
    target_df['len_type'] = target_df['comment_content'].apply(categorize_length)
    
    # 3. 分别计算三类的绝对数量 (Absolute Volume)
    # 我们不看比例，只看绝对量，因为我们已经吸取了之前的教训
    short_vol = target_df[target_df['len_type'] == 'Short']['comment_content'].resample(RESAMPLE_FREQ).count()
    medium_vol = target_df[target_df['len_type'] == 'Medium']['comment_content'].resample(RESAMPLE_FREQ).count()
    long_vol = target_df[target_df['len_type'] == 'Long']['comment_content'].resample(RESAMPLE_FREQ).count()
    
    # 4. 总量 (用于背景参考)
    total_vol = target_df['comment_content'].resample(RESAMPLE_FREQ).count()

    # --- 绘图 ---
    fig, ax1 = plt.subplots(figsize=(14, 7))

    # 背景：总量 (灰色填充)
    ax1.fill_between(total_vol.index, 0, total_vol.values, color='gainsboro', alpha=0.5, label='总流量')

    # 线条：三类文本的走势
    # 短文本：红色 (情绪化)
    ax1.plot(short_vol.index, short_vol.values, 
             color='#d62728', linestyle='-', linewidth=2, label='短文本 (<10字): 宣泄/复读')
    
    # 长文本：蓝色 (理智/深情) - 这是你要验证的核心
    ax1.plot(long_vol.index, long_vol.values, 
             color='#1f77b4', linestyle='-', marker='*', linewidth=2.5, label='长文本 (>50字): 小作文/维权')
    
    # 中文本：绿色 (日常)
    ax1.plot(medium_vol.index, medium_vol.values, 
             color='#2ca02c', linestyle=':', alpha=0.6, label='中文本 (10-50字): 日常讨论')

    # 设置 Y 轴
    ax1.set_ylabel('评论条数 (Absolute Count)', fontsize=12)
    ax1.set_xlabel('时间')
    
    # 格式化
    ax1.xaxis.set_major_formatter(DateFormatter("%m-%d %H:%M"))
    ax1.legend(loc='upper left')
    
    plt.title(f"验证10: 舆情结构分析 — 是谁在主导舆论？", fontsize=14)
    plt.tight_layout()
    plt.show()

    # --- 打印分析 ---
    max_long_time = long_vol.idxmax()
    max_long_val = long_vol.max()
    print(f"\n📝 文本结构报告:")
    print(f"   - '小作文'爆发时刻: {max_long_time}")
    print(f"   - 最高单周期长评数: {max_long_val} 条")
    if max_long_val > 10: # 假设阈值
        print("   - 🚨 判定: 核心玩家深度破防，正在进行长文维权。")
```

## 验证情感极性&语义距离漂移

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.dates import DateFormatter
from sentence_transformers import SentenceTransformer, util
from transformers import pipeline
import torch
from tqdm.auto import tqdm  # 使用 auto 可以自动适配 Notebook 或终端

# 引用你现有的工具
from utilis_preprocess import batch_clean_text

# ==========================================
# 1. 严格清洗函数 (提纯)
# ==========================================
def strict_clean_data(df, official_source, irrelevant_source):
    """
    数据提纯：
    1. 基础清洗 (batch_clean_text)
    2. 严格剔除 FILTER_NEUTRAL_LEACHING
    3. 剔除过短文本 (再次确认)
    """
    print("🧹 正在进行严格数据提纯 (去除中性/无关噪音)...")
    
    # 1. 复用你现有的清洗逻辑
    # 注意：这里使用 "full" 模式以获得更干净的文本
    df = batch_clean_text(df, "simple", official_source, irrelevant_source)
    
    # 2. 核心：剔除被标记为噪音的数据
    # 这一步是把你觉得“影响结论”的中性数据干掉的关键
    df = df[df['text_clean'] != "FILTER_NEUTRAL_LEACHING"].copy()
    
    # 3. 额外过滤：剔除纯数字或极短的文本 (往往是无意义的)
    df = df[df['text_clean'].str.len() > 3].copy()
    
    print(f"✅ 提纯完成，剩余有效样本: {len(df)} 条")
    return df

# ==========================================
# 2. 验证 V11: 情感极性趋势 (Sentiment Polarity)
# ==========================================
def verify_sentiment_trend(df, start_date, end_date, model_name="uer/roberta-base-finetuned-dianping-chinese"):
    print(f"\n🧪 正在验证 V11: 情感极性趋势 (Sentiment Polarity)...")
    
    # 时间切片
    df['dt'] = pd.to_datetime(df['timestamp'], errors='coerce')
    mask = (df['dt'] >= start_date) & (df['dt'] <= end_date)
    target_df = df.loc[mask].copy()
    
    if target_df.empty: 
        print("❌ 该时间段无数据")
        return

    # --- A. 实时推理情感 (如果数据里没有 sentiment_score) ---
    if 'sentiment_score' not in target_df.columns:
        print(f"   - 未检测到 sentiment_score，正在调用模型 ({model_name}) 进行推理...")
        
        device = 0 if torch.cuda.is_available() else -1
        sentiment_pipeline = pipeline(
            "text-classification", 
            model=model_name, 
            device=device, 
            batch_size=64, 
            truncation=True, 
            max_length=512
        )
        
        texts = target_df['text_clean'].tolist()
        results = []
        batch_size = 64
        
        # 【修改点】加入 tqdm 进度条
        for i in tqdm(range(0, len(texts), batch_size), desc="   - 情感推理进度", unit="batch"):
            batch = texts[i:i+batch_size]
            results.extend(sentiment_pipeline(batch))
        
        # 提取分数 (假设 label 包含 negative/positive 或 0/1)
        scores = []
        for r in results:
            s = r['score']
            label = str(r['label']).lower()
            # 逻辑：如果是负面标签，则分数转为 (1-s)，即“正向概率”变低
            # 注意：不同模型的 label 定义不同，这里假设 'positive'/'1' 是正向
            if 'negative' in label or label == '0' or ('1' in label and s < 0.5): 
                s = 1 - s # 转化为正向概率
            scores.append(s)
        target_df['sentiment_score'] = scores
    
    target_df.set_index('dt', inplace=True)
    RESAMPLE_FREQ = '15T'

    # --- B. 计算指标 ---
    # 1. 流量
    vol_series = target_df['text_clean'].resample(RESAMPLE_FREQ).count()
    
    # 2. 情感均值 (Mean Sentiment)
    sent_series = target_df['sentiment_score'].resample(RESAMPLE_FREQ).mean()
    
    # 3. 负面占比 (Negative Ratio) - 设定阈值 < 0.4 为负面
    def calc_neg_ratio(x):
        if len(x) == 0: return 0
        return (x < 0.4).sum() / len(x)
    
    neg_ratio_series = target_df['sentiment_score'].resample(RESAMPLE_FREQ).apply(calc_neg_ratio)

    # --- C. 绘图 ---
    fig, ax1 = plt.subplots(figsize=(14, 6))

    # 背景：流量
    ax1.bar(vol_series.index, vol_series.values, width=0.015, color='gainsboro', label='总流量', alpha=0.6)
    ax1.set_ylabel('流量 (Volume)', color='gray')
    ax1.set_xlabel('时间')

    # 右轴：情感指标
    ax2 = ax1.twinx()
    
    # 线1: 情感均值 (蓝色) - 越低越负面
    ax2.plot(sent_series.index, sent_series.values, 
             color='#1f77b4', linestyle='-', linewidth=2, marker='.', label='情感均值 (Mean Score)')
    
    # 线2: 负面占比 (红色) - 越高越危险
    ax2.plot(neg_ratio_series.index, neg_ratio_series.values,
             color='#d62728', linestyle='--', linewidth=2, label='负面评论占比 (Neg Ratio)')
    
    # 警戒线
    ax2.axhline(y=0.4, color='orange', linestyle=':', label='均值警戒线 (0.4)')
    
    ax2.set_ylabel('情感指标 (0=负, 1=正)', color='#333')
    ax2.set_ylim(0, 1.1)
    
    ax1.xaxis.set_major_formatter(DateFormatter("%m-%d %H:%M"))
    
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax2.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    
    plt.title(f"验证11: 情感极性验证 (已剔除中性噪音)", fontsize=14)
    plt.tight_layout()
    plt.show()

# ==========================================
# 3. 验证 V12: 语义漂移 (Semantic Shift)
# ==========================================
def verify_semantic_shift(df, start_date, end_date, model_name="paraphrase-multilingual-MiniLM-L12-v2"):
    print(f"\n🧪 正在验证 V12: 语义漂移检测 (Semantic Shift)...")
    
    # 初始化模型
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"   - Loading Embedding Model on {device}...")
    embedder = SentenceTransformer(model_name, device=device)
    
    # 数据切片
    df['dt'] = pd.to_datetime(df['timestamp'], errors='coerce')
    mask = (df['dt'] >= start_date) & (df['dt'] <= end_date)
    target_df = df.loc[mask].copy()
    
    if target_df.empty: 
        print("❌ 该时间段无数据")
        return

    # 按时间窗口聚合文本
    RESAMPLE_FREQ = '15T' # 建议用 1小时
    grouped = target_df.groupby(pd.Grouper(key='dt', freq=RESAMPLE_FREQ))
    
    timeline = []
    volumes = []
    embeddings_map = {}

    # 【修改点】优化 tqdm 描述
    print("   - 正在计算每小时的语义中心向量...")
    for time_bucket, group in tqdm(grouped, desc="   - 语义向量计算", unit="bucket"):
        if group.empty:
            timeline.append(time_bucket)
            volumes.append(0)
            embeddings_map[time_bucket] = None
            continue
            
        texts = group['text_clean'].tolist()
        # 编码该时间段的所有文本
        # 注意：这里 show_progress_bar=False 是为了不和外层 tqdm 冲突
        vectors = embedder.encode(texts, show_progress_bar=False, batch_size=64)
        center_vec = np.mean(vectors, axis=0)
        
        timeline.append(time_bucket)
        volumes.append(len(texts))
        embeddings_map[time_bucket] = center_vec
    
    # 计算漂移 (当前时刻 vs 上一时刻 的余弦距离)
    shifts = []
    prev_vec = None
    
    for t in timeline:
        curr_vec = embeddings_map[t]
        
        if curr_vec is None or prev_vec is None:
            shifts.append(0.0) # 无法计算，漂移为0
        else:
            # Cosine Distance = 1 - Cosine Similarity
            sim = util.cos_sim(curr_vec, prev_vec).item()
            dist = 1 - sim
            shifts.append(dist)
        
        if curr_vec is not None:
            prev_vec = curr_vec
            
    # --- 绘图 ---
    fig, ax1 = plt.subplots(figsize=(14, 6))

    # 背景：流量
    ax1.bar(timeline, volumes, width=0.03, color='gainsboro', label='总流量', alpha=0.6)
    ax1.set_ylabel('流量', color='gray')
    
    # 右轴：语义漂移
    ax2 = ax1.twinx()
    color_shift = '#9467bd' # 紫色
    
    ax2.plot(timeline, shifts, color=color_shift, marker='x', linestyle='-', linewidth=2, label='语义漂移度 (Semantic Shift)')
    
    # 漂移警戒线
    ax2.axhline(y=0.1, color=color_shift, linestyle=':', alpha=0.5, label='变题警戒线')
    
    ax2.set_ylabel('语义距离 (1 - Cosine Sim)', color=color_shift)
    ax2.set_ylim(0, 0.5) 
    
    ax1.xaxis.set_major_formatter(DateFormatter("%m-%d %H:%M"))
    
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax2.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    
    plt.title(f"验证12: 语义漂移检测 — 舆论焦点是否发生突变？", fontsize=14)
    plt.tight_layout()
    plt.show()
    
    # 打印最大漂移时刻
    max_shift_idx = np.argmax(shifts)
    print(f"\n🌪️ 语义分析报告:")
    print(f"   - 话题突变最剧烈的时刻: {timeline[max_shift_idx]}")
    print(f"   - 漂移指数: {shifts[max_shift_idx]:.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.dates import DateFormatter
from matplotlib.lines import Line2D
from utilis_preprocess import parse_weibo_date

def verify_official_interaction_v2(user_df, official_df, start_date, end_date):
    print(f"📢 正在验证 V13-Pro: 官方全类型行为影响分析...")
    
    # --- 1. 数据预处理 ---
    # 用户数据
    user_df['dt'] = pd.to_datetime(user_df['timestamp'], errors='coerce')
    mask = (user_df['dt'] >= start_date) & (user_df['dt'] <= end_date)
    u_df = user_df.loc[mask].copy()
    
    # 官方数据清洗
    if 'created_at' in official_df.columns:
        official_df.rename(columns={'created_at': 'timestamp'}, inplace=True)
    official_df['dt'] = pd.to_datetime(official_df['timestamp'].apply(parse_weibo_date), errors='coerce')
    o_mask = (official_df['dt'] >= start_date) & (official_df['dt'] <= end_date)
    o_df = official_df.loc[o_mask].copy()
    
    if u_df.empty: 
        print("❌ 用户数据为空")
        return

    RESAMPLE_FREQ = '30T'
    
    # --- 2. 应用你定义的映射字典 ---
    type_mapping = {
        # 宣发类 (蓝)
        '版本宣发': 'promotion', '其他活动': 'promotion', '抽奖': 'promotion', 'PV': 'promotion',
        
        # 危机/道歉类 (红) - "维护&道歉" 归入此类，因为带有负面承认
        '道歉公告': 'crisis', '维护&道歉': 'crisis', '致歉': 'crisis',
        
        # 维护类 (黄)
        '更新公告': 'maintenance', '维护公告': 'maintenance',
        
        # 攻略/干货类 (绿) - 新增
        '官方攻略': 'guide', '搭配指南': 'guide',
        
        # 日常 (灰)
        '日常互动': 'daily'
    }
    
    # 如果 official_post_type 列存在，进行映射；否则尝试从文本推断
    if 'official_post_type' in o_df.columns:
        o_df['category'] = o_df['official_post_type'].map(type_mapping).fillna('daily')
    else:
        print("⚠️ 未找到 official_post_type 列，尝试使用关键词模糊匹配...")
        def fuzzy_map(text):
            text = str(text)
            if '道歉' in text or '修复' in text: return 'crisis'
            if '攻略' in text: return 'guide'
            if '抽奖' in text or '宣发' in text: return 'promotion'
            if '维护' in text or '更新' in text: return 'maintenance'
            return 'daily'
        o_df['category'] = o_df['text'].apply(fuzzy_map)

    # --- 3. 计算用户侧指标 ---
    # A. 总流量
    # vol_series = u_df.set_index('dt')['text_clean'].resample(RESAMPLE_FREQ).count()
    vol_series = u_df.set_index('dt')['content'].resample(RESAMPLE_FREQ).count()

    
    # B. 原创占比 (Original Ratio)
    # 逻辑：攻略和宣发通常转发率高(原创率低)；危机通常原创率高
    if 'source_post_id' in u_df.columns:
        orig_ratio_series = u_df.set_index('dt')['source_post_id'].resample(RESAMPLE_FREQ).apply(
            lambda x: x.isna().sum() / (len(x) + 1e-5)
        )
    else:
        orig_ratio_series = pd.Series(0.5, index=vol_series.index) # 默认值

    # --- 4. 绘图 ---
    fig, ax1 = plt.subplots(figsize=(15, 8))

    # 背景：用户流量 (灰色填充)
    ax1.fill_between(vol_series.index, 0, vol_series.values, color='gainsboro', alpha=0.6, label='用户流量')
    ax1.set_ylabel('用户活跃度 (Volume)', color='gray', fontsize=12)
    ax1.set_xlabel('时间')

    # 右轴：原创占比
    ax2 = ax1.twinx()
    ax2.plot(orig_ratio_series.index, orig_ratio_series.values, 
             color='#1f77b4', linestyle='-', linewidth=2, alpha=0.8, label='原创微博占比')
    
    # 辅助线
    ax2.axhline(y=0.2, color='green', linestyle=':', alpha=0.5, label='高转发特征线 (<0.2)')
    ax2.axhline(y=0.8, color='red', linestyle=':', alpha=0.5, label='高原创特征线 (>0.8)')
    ax2.set_ylabel('交互模式 (0=转发狂欢, 1=原创怒喷)', color='#1f77b4', fontsize=12)
    ax2.set_ylim(0, 1.1)

    # --- 5. 叠加官方动作 (垂直线) ---
    # 获取最大流量值用于定位标签高度
    y_max = vol_series.max() if not vol_series.empty else 10

    # A. 宣发 (蓝色虚线)
    promos = o_df[o_df['category'] == 'promotion']
    for t in promos['dt']:
        ax1.axvline(x=t, color='blue', linestyle='--', alpha=0.6, ymax=0.4)
        ax1.text(t, y_max*0.42, '宣', color='blue', ha='center', fontsize=9)

    # B. 危机/道歉 (红色实线) - 包含 "维护&道歉"
    crises = o_df[o_df['category'] == 'crisis']
    for t in crises['dt']:
        ax1.axvline(x=t, color='red', linestyle='-', linewidth=2, alpha=0.8, ymax=0.6)
        ax1.text(t, y_max*0.62, '歉', color='red', ha='center', fontweight='bold', fontsize=10)
        
    # C. 攻略 (绿色点划线) - 新增
    guides = o_df[o_df['category'] == 'guide']
    for t in guides['dt']:
        ax1.axvline(x=t, color='green', linestyle='-.', alpha=0.8, ymax=0.3)
        ax1.text(t, y_max*0.32, '略', color='green', ha='center', fontsize=9)

    # D. 维护 (橙色点线)
    maints = o_df[o_df['category'] == 'maintenance']
    for t in maints['dt']:
        ax1.axvline(x=t, color='orange', linestyle=':', alpha=0.8, ymax=0.2)
        ax1.text(t, y_max*0.22, '维', color='orange', ha='center', fontsize=8)

    # 格式化
    ax1.xaxis.set_major_formatter(DateFormatter("%m-%d %H:%M"))
    plt.title(f"验证13-Pro: 官方全行为激励验证\n(含 '维护&道歉' 与 '攻略' 识别)", fontsize=14)
    
    # 自定义图例
    custom_lines = [
        Line2D([0], [0], color='blue', linestyle='--', label='宣发/抽奖'),
        Line2D([0], [0], color='red', linestyle='-', lw=2, label='道歉/修复'),
        Line2D([0], [0], color='green', linestyle='-.', label='官方攻略'),
        Line2D([0], [0], color='orange', linestyle=':', label='常规维护'),
        Line2D([0], [0], color='#1f77b4', linestyle='-', label='用户原创占比')
    ]
    ax1.legend(handles=custom_lines, loc='upper left')
    
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.dates import DateFormatter
from matplotlib.lines import Line2D
from utilis_preprocess import parse_weibo_date
from version_config import OFFICIAL

def verify_official_coupling(user_df, comment_df,official_df, start_date, end_date, official_uid_list=OFFICIAL):
    """
    V14: 官方耦合度分析
    official_uid_list: 官方账号的 user_id 列表 (例如 ['123456', '654321'])
    """
    print(f"🔗 正在验证 V14: 官方交互耦合度 — 流量是'聚拢'还是'离散'？...")
    
    # --- 1. 数据预处理 ---
    user_df['dt'] = pd.to_datetime(user_df['timestamp'], errors='coerce')
    mask = (user_df['dt'] >= start_date) & (user_df['dt'] <= end_date)
    u_df = user_df.loc[mask].copy()

    comment_df['dt'] = pd.to_datetime(comment_df['timestamp'], errors='coerce')
    mask = (comment_df['dt'] >= start_date) & (comment_df['dt'] <= end_date)
    c_df = comment_df.loc[mask].copy()
    
    # 官方数据清洗
    if 'created_at' in official_df.columns:
        official_df.rename(columns={'created_at': 'timestamp'}, inplace=True)
    official_df['dt'] = pd.to_datetime(official_df['timestamp'].apply(parse_weibo_date), errors='coerce')
    o_mask = (official_df['dt'] >= start_date) & (official_df['dt'] <= end_date)
    o_df = official_df.loc[o_mask].copy()
    
    if u_df.empty: return

    RESAMPLE_FREQ = '15T'
    
    # --- 2. 关键计算：区分“官方下”和“广场上” ---
    
    # 确保 id 列是字符串类型，防止匹配失败
    # 假设列名是 'root_uid' 或 '评论对应微博的author_id' (根据你的描述调整)
    target_col = None
    possible_cols = ['official_author', 'source_author_name'] # 增加一些备选
    for col in possible_cols:
        if col in u_df.columns:
            target_col = col
            break
            
    if target_col is None:
        print(f"❌ 错误：在用户数据中找不到 '评论目标作者ID' 相关列，无法计算耦合度。")
        print(f"   现有列名: {u_df.columns.tolist()}")
        return

    # 转换为字符串列表以便匹配
    str_uid_list = [str(uid) for uid in official_uid_list]
    
    # A.1. 官方流量 (Official Volume): 目标是官方ID的评论
    is_official_target = u_df['source_author_name'].astype(str).isin(str_uid_list)
    official_vol_p = u_df[is_official_target].set_index('dt')['content'].resample(RESAMPLE_FREQ).count()


    is_official_target = c_df['official_author'].astype(str).isin(str_uid_list)
    official_vol_c = c_df[is_official_target].set_index('dt')['comment_content'].resample(RESAMPLE_FREQ).count()

    official_vol = official_vol_c+official_vol_p
    
    # B. 总流量 (Total Volume)
    total_vol = u_df.set_index('dt')['content'].resample(RESAMPLE_FREQ).count()+c_df[is_official_target].set_index('dt')['comment_content'].resample(RESAMPLE_FREQ).count()
    
    # C. 耦合度 (Coupling Ratio)
    # 避免除以零
    coupling_ratio = (official_vol_p+official_vol_c) / (total_vol + 1e-5)

    # --- 3. 官方动作映射 (复用之前的逻辑) ---
    type_mapping = {
        '版本宣发': 'promotion', '其他活动': 'promotion', '抽奖': 'promotion',
        '道歉公告': 'crisis', '维护&道歉': 'crisis', 
        '官方攻略': 'guide',
        '更新公告': 'maintenance', '维护公告': 'maintenance',
        '日常互动': 'daily'
    }
    if 'official_post_type' in o_df.columns:
        o_df['category'] = o_df['official_post_type'].map(type_mapping).fillna('daily')
    else:
        o_df['category'] = 'daily' # 简化处理

    # --- 4. 绘图 ---
    fig, ax1 = plt.subplots(figsize=(15, 8))

    # 背景1：总流量 (浅灰色)
    ax1.fill_between(total_vol.index, 0, total_vol.values, color='lightgray', alpha=0.5, label='社区总流量 (广场+官方)')
    
    # 背景2：官方下流量 (深灰色) -> 让你直观看到有多少是直接冲着官方去的
    ax1.fill_between(official_vol.index, 0, official_vol.values, color='dimgray', alpha=0.6, label='官方评论区流量')
    
    ax1.set_ylabel('流量规模', color='gray')

    # 右轴：耦合度
    ax2 = ax1.twinx()
    # 颜色逻辑：耦合度高(紫色) = 官方控场强；耦合度低(黑色) = 社区失控
    ax2.plot(coupling_ratio.index, coupling_ratio.values, 
             color='#9467bd', linestyle='-', linewidth=2.5, marker='.', markersize=5, label='官方耦合度 (Ratio)')
    
    ax2.set_ylabel('官方耦合度 (0=完全脱离, 1=完全聚集)', color='#9467bd')
    ax2.set_ylim(0, 1.1)
    
    # 辅助线
    ax2.axhline(y=0.5, color='#9467bd', linestyle=':', alpha=0.5, label='50% 警戒线')

    # --- 5. 叠加官方动作 ---
    y_max = total_vol.max() if not total_vol.empty else 10
    
    # 宣发 (蓝)
    for t in o_df[o_df['category'] == 'promotion']['dt']:
        ax1.axvline(x=t, color='blue', linestyle='--', alpha=0.5)
        ax1.text(t, y_max*0.9, '宣', color='blue', ha='center')

    # 危机 (红)
    for t in o_df[o_df['category'] == 'crisis']['dt']:
        ax1.axvline(x=t, color='red', linestyle='-', linewidth=2, alpha=0.8)
        ax1.text(t, y_max*0.95, '歉', color='red', ha='center', fontweight='bold')
        
    # 攻略 (绿)
    for t in o_df[o_df['category'] == 'guide']['dt']:
        ax1.axvline(x=t, color='green', linestyle='-.', alpha=0.6)
        ax1.text(t, y_max*0.85, '略', color='green', ha='center')

    ax1.xaxis.set_major_formatter(DateFormatter("%m-%d %H:%M"))
    
    # 图例
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    
    plt.title(f"验证14: 官方耦合度分析 — 舆论是否在官方的掌控之中？", fontsize=14)
    plt.tight_layout()
    plt.show()

# --- 调用示例 ---
# ⚠️ 请替换为你实际的官方 UID 列表 (字符串格式)
# 例如无限暖暖官方微博 UID
# user_df 必须包含 '评论对应微博的author_id' 这一列
# verify_official_coupling(df_c, df_official, '2024-11-01', '2024-11-02', ['官方UID1', '官方UID2'])
```

```python
"""
视觉重复度检测 - 精简版 V3
文件命名规则: md5(url).jpg
"""

import os
import pickle
import hashlib
from pathlib import Path
from collections import defaultdict

import numpy as np
import pandas as pd
from PIL import Image
import imagehash
from tqdm import tqdm
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False


class VisualRepetitionAnalyzer:
    """视觉重复度分析器"""
    
    def __init__(self, 
                 image_base_dir: str = "./weibo_images",
                 cache_path: str = "./phash_cache.pkl",
                 hamming_threshold: int = 10):
        self.image_base_dir = Path(image_base_dir)
        self.cache_path = cache_path
        self.hamming_threshold = hamming_threshold
        self.hash_cache = self._load_cache()
    
    def _load_cache(self) -> dict:
        if os.path.exists(self.cache_path):
            try:
                with open(self.cache_path, 'rb') as f:
                    cache = pickle.load(f)
                    print(f"✅ 加载 {len(cache)} 条哈希缓存")
                    return cache
            except:
                pass
        return {}
    
    def save_cache(self):
        with open(self.cache_path, 'wb') as f:
            pickle.dump(self.hash_cache, f)
        print(f"💾 保存 {len(self.hash_cache)} 条缓存")
    
    # =========================================================================
    # URL解析 - 使用 MD5
    # =========================================================================
    
    def parse_image_field(self, field) -> list:
        """解析图片字段 - 逗号+空格分隔"""
        if pd.isna(field) or field == '':
            return []
        return str(field).split(', ')
    
    def url_to_local_path(self, url: str) -> str | None:
        """
        URL转本地路径
        规则: filename = md5(url).jpg, path = base/xx/filename
        """
        if not url or pd.isna(url):
            return None
        
        url = str(url).strip()
        if not url:
            return None
        
        # 计算 URL 的 MD5
        filename = hashlib.md5(url.encode('utf-8')).hexdigest() + '.jpg'
        local_path = self.image_base_dir / filename[:2] / filename
        
        return str(local_path) if local_path.exists() else None
    
    # =========================================================================
    # 哈希与聚类
    # =========================================================================
    
    def compute_phash(self, path: str) -> str | None:
        if path in self.hash_cache:
            return self.hash_cache[path]
        try:
            img = Image.open(path).convert('RGB')
            h = str(imagehash.phash(img, hash_size=16))
            self.hash_cache[path] = h
            return h
        except:
            return None
    
    def cluster_images(self, hash_dict: dict) -> list:
        """Union-Find聚类相似图片"""
        if not hash_dict:
            return []
        
        paths = list(hash_dict.keys())
        n = len(paths)
        parent = list(range(n))
        
        def find(x):
            if parent[x] != x:
                parent[x] = find(parent[x])
            return parent[x]
        
        def union(x, y):
            px, py = find(x), find(y)
            if px != py:
                parent[py] = px
        
        hashes = [hash_dict[p] for p in paths]
        for i in range(n):
            for j in range(i + 1, n):
                try:
                    dist = imagehash.hex_to_hash(hashes[i]) - imagehash.hex_to_hash(hashes[j])
                    if dist <= self.hamming_threshold:
                        union(i, j)
                except:
                    pass
        
        clusters = defaultdict(list)
        for i in range(n):
            clusters[find(i)].append(paths[i])
        return list(clusters.values())
    
    def calc_metrics(self, hash_dict: dict) -> dict:
        """计算重复率指标"""
        if not hash_dict:
            return {'total': 0, 'unique': 0, 'rep_rate': np.nan, 
                    'max_cluster': 0, 'max_ratio': 0}
        
        clusters = self.cluster_images(hash_dict)
        total = len(hash_dict)
        unique = len(clusters)
        max_c = max(len(c) for c in clusters) if clusters else 0
        
        return {
            'total': total,
            'unique': unique,
            'rep_rate': 1 - unique / total if total > 0 else 0,
            'max_cluster': max_c,
            'max_ratio': max_c / total if total > 0 else 0
        }
    
    # =========================================================================
    # 时序分析
    # =========================================================================
    
    def analyze(self, df: pd.DataFrame, start_date: str, end_date: str,
                image_col: str = 'image_urls', video_col: str = 'video_cover_url',
                time_col: str = 'timestamp', freq: str = '15T') -> pd.DataFrame:
        
        print(f"🔍 分析: {start_date} ~ {end_date}")
        
        # 过滤时间
        df = df.copy()
        df['_dt'] = pd.to_datetime(df[time_col], errors='coerce')
        df = df[(df['_dt'] >= start_date) & (df['_dt'] <= end_date)]
        
        if df.empty:
            print("⚠️ 无数据")
            return pd.DataFrame()
        
        print(f"   数据量: {len(df)} 条")
        
        # 解析路径
        print("📁 解析图片路径 (MD5映射)...")
        
        def get_paths(row):
            paths = []
            if image_col in row.index:
                for url in self.parse_image_field(row[image_col]):
                    p = self.url_to_local_path(url)
                    if p: 
                        paths.append(p)
            if video_col in row.index:
                p = self.url_to_local_path(row[video_col])
                if p: 
                    paths.append(p)
            return paths
        
        df['_paths'] = df.apply(get_paths, axis=1)
        df['_has_img'] = df['_paths'].apply(lambda x: 1 if x else 0)
        
        total_found = sum(len(p) for p in df['_paths'])
        print(f"   找到 {total_found} 张本地图片")
        
        if total_found == 0:
            print("⚠️ 未找到图片")
            return pd.DataFrame()
        
        # 预计算哈希
        print("🔢 计算pHash...")
        all_paths = set(p for paths in df['_paths'] for p in paths)
        new_paths = [p for p in all_paths if p not in self.hash_cache]
        
        if new_paths:
            for p in tqdm(new_paths, desc="计算哈希"):
                self.compute_phash(p)
            self.save_cache()
        
        # 按窗口分析
        print(f"📊 按 {freq} 窗口统计...")
        df.set_index('_dt', inplace=True)
        results = []
        
        for t, g in tqdm(list(df.groupby(pd.Grouper(freq=freq))), desc="分析"):
            if g.empty: 
                continue
            
            paths = [p for ps in g['_paths'] for p in ps]
            hashes = {p: self.hash_cache[p] for p in paths if p in self.hash_cache}
            m = self.calc_metrics(hashes)
            
            results.append({
                'time': t,
                'total_volume': len(g),
                'with_img_count': int(g['_has_img'].sum()),
                'with_img_ratio': g['_has_img'].mean(),
                'img_count': m['total'],
                'unique_groups': m['unique'],
                'rep_rate': m['rep_rate'],
                'max_cluster': m['max_cluster'],
                'max_ratio': m['max_ratio']
            })
        
        result_df = pd.DataFrame(results)
        self._summary(result_df)
        return result_df
    
    def _summary(self, df):
        if df.empty: 
            return
        print(f"\n{'='*50}")
        print(f"📋 {len(df)} 窗口 | 帖子 {df['total_volume'].sum()} | "
              f"含图 {df['with_img_count'].sum()} ({df['with_img_ratio'].mean():.1%})")
        valid = df['rep_rate'].dropna()
        if len(valid) > 0:
            print(f"   平均重复率 {valid.mean():.1%} | 最高 {valid.max():.1%} | 最大簇 {df['max_cluster'].max()}")
        print('='*50)


def plot_visual_analysis(df: pd.DataFrame, save_path: str = None):
    """可视化"""
    if df.empty: 
        return
    
    fig, axes = plt.subplots(2, 2, figsize=(15, 9))
    t = df['time']
    
    # ① 总量 + 重复率
    ax = axes[0, 0]
    ax.bar(t, df['total_volume'], width=0.008, color='#1f77b4', alpha=0.4)
    ax.set_ylabel('发帖量', color='#1f77b4')
    ax2 = ax.twinx()
    ax2.plot(t, df['rep_rate'], 'r-o', markersize=3, lw=1.5)
    ax2.axhline(0.3, color='orange', ls='--', alpha=0.7)
    ax2.axhline(0.5, color='red', ls='--', alpha=0.7)
    ax2.set_ylabel('重复率', color='red')
    ax2.set_ylim(0, 1.05)
    ax.set_title('① 发帖量 & 视觉重复率', fontweight='bold')
    
    # ② 含图量 + 含图比例
    ax = axes[0, 1]
    ax.bar(t, df['with_img_count'], width=0.008, color='#2ca02c', alpha=0.5)
    ax.set_ylabel('含图数', color='#2ca02c')
    ax2 = ax.twinx()
    ax2.plot(t, df['with_img_ratio']*100, 'purple', marker='s', markersize=3)
    ax2.set_ylabel('含图比例%', color='purple')
    ax2.set_ylim(0, 105)
    ax.set_title('② 含图量 & 含图比例', fontweight='bold')
    
    # ③ 最大重复簇
    ax = axes[1, 0]
    colors = ['red' if x > 5 else 'green' for x in df['max_cluster']]
    ax.bar(t, df['max_cluster'], width=0.008, color=colors, alpha=0.7)
    ax.axhline(5, color='orange', ls='--')
    ax.set_ylabel('最大簇')
    ax.set_title('③ 最大相同图片簇', fontweight='bold')
    
    # ④ 独立组 vs 总图片
    ax = axes[1, 1]
    ax.fill_between(t, 0, df['img_count'], alpha=0.3, label='总图片')
    ax.fill_between(t, 0, df['unique_groups'], alpha=0.6, color='green', label='独立组')
    ax.legend(loc='upper right')
    ax.set_title('④ 视觉多样性', fontweight='bold')
    
    for ax in axes.flat:
        ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
        plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right', fontsize=8)
    
    plt.suptitle('视觉同质性检测报告', fontsize=14, fontweight='bold')
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"📊 已保存: {save_path}")
    plt.show()


def verify_visual_impact(df, start_date, end_date, 
                         image_base_dir="./weibo_images",
                         freq='15T', hamming_threshold=10, save_plot=None):
    """主入口"""
    analyzer = VisualRepetitionAnalyzer(image_base_dir, hamming_threshold=hamming_threshold)
    result = analyzer.analyze(df, start_date, end_date, freq=freq)
    if not result.empty:
        plot_visual_analysis(result, save_plot)
    return result
```

### 关键函数

```python
import hashlib
def md5(text: str) -> str:
    """
    计算给定字符串的 MD5 散列值。

    参数:
        text (str): 要计算 MD5 的输入字符串。

    返回:
        str: 字符串的十六进制 MD5 散列值。
    """
    # 1. 创建 MD5 对象
    m = hashlib.md5()
    
    # 2. 将字符串编码为 bytes 并更新 MD5 对象
    #    .encode('utf-8') 将字符串转换为 bytes
    m.update(text.encode('utf-8'))
    
    # 3. 获取十六进制散列值
    return m.hexdigest()
```

## 图片重复率V3

```python
"""
视觉重复度检测 - 增强版 V4
新增：组合特征检测事件、异常检测、持续聚类分析
"""

import os
import pickle
import hashlib
from pathlib import Path
from collections import defaultdict

import numpy as np
import pandas as pd
from PIL import Image
import imagehash
from tqdm import tqdm
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False


class VisualRepetitionAnalyzer:
    """视觉重复度分析器 - 增强版"""
    
    def __init__(self, 
                 image_base_dir: str = IMAGE_PATH,
                 cache_path: str = "./phash_cache.pkl",
                 hamming_threshold: int = 10,
                 min_volume_threshold: int = 20):  # 新增：最小图片量阈值
        self.image_base_dir = Path(image_base_dir)
        self.cache_path = cache_path
        self.hamming_threshold = hamming_threshold
        self.min_volume_threshold = min_volume_threshold
        self.hash_cache = self._load_cache()
    
    def _load_cache(self) -> dict:
        if os.path.exists(self.cache_path):
            try:
                with open(self.cache_path, 'rb') as f:
                    cache = pickle.load(f)
                    print(f"✅ 加载 {len(cache)} 条哈希缓存")
                    return cache
            except:
                pass
        return {}
    
    def save_cache(self):
        with open(self.cache_path, 'wb') as f:
            pickle.dump(self.hash_cache, f)
        print(f"💾 保存 {len(self.hash_cache)} 条缓存")
    
    # =========================================================================
    # URL解析
    # =========================================================================
    
    def parse_image_field(self, field) -> list:
        if pd.isna(field) or field == '':
            return []
        return str(field).split(', ')
    
    def url_to_local_path(self, url: str) -> str | None:
        if not url or pd.isna(url):
            return None
        url = str(url).strip()
        if not url:
            return None
        filename = hashlib.md5(url.encode('utf-8')).hexdigest() + '.jpg'
        local_path = self.image_base_dir / filename[:2] / filename
        return str(local_path) if local_path.exists() else None
    
    # =========================================================================
    # 哈希与聚类
    # =========================================================================
    
    def compute_phash(self, path: str) -> str | None:
        if path in self.hash_cache:
            return self.hash_cache[path]
        try:
            img = Image.open(path).convert('RGB')
            h = str(imagehash.phash(img, hash_size=16))
            self.hash_cache[path] = h
            return h
        except:
            return None
    
    def cluster_images(self, hash_dict: dict) -> list:
        """Union-Find聚类，返回按大小降序排列的簇"""
        if not hash_dict:
            return []
        
        paths = list(hash_dict.keys())
        n = len(paths)
        parent = list(range(n))
        
        def find(x):
            if parent[x] != x:
                parent[x] = find(parent[x])
            return parent[x]
        
        def union(x, y):
            px, py = find(x), find(y)
            if px != py:
                parent[py] = px
        
        hashes = [hash_dict[p] for p in paths]
        for i in range(n):
            for j in range(i + 1, n):
                try:
                    dist = imagehash.hex_to_hash(hashes[i]) - imagehash.hex_to_hash(hashes[j])
                    if dist <= self.hamming_threshold:
                        union(i, j)
                except:
                    pass
        
        clusters = defaultdict(list)
        for i in range(n):
            clusters[find(i)].append(paths[i])
        
        # 按大小降序排列
        return sorted(clusters.values(), key=len, reverse=True)
    
    def calc_metrics(self, hash_dict: dict) -> dict:
        """计算基础+增强指标"""
        if not hash_dict:
            return {
                'total': 0, 'unique': 0, 'rep_rate': np.nan,
                'max_cluster': 0, 'max_ratio': 0,
                'top3_cluster_sum': 0, 'cluster_sizes': []
            }
        
        clusters = self.cluster_images(hash_dict)
        total = len(hash_dict)
        unique = len(clusters)
        cluster_sizes = [len(c) for c in clusters]
        max_c = cluster_sizes[0] if cluster_sizes else 0
        
        # Top-K 簇总和 (取前3大簇)
        top3_sum = sum(cluster_sizes[:3])
        
        return {
            'total': total,
            'unique': unique,
            'rep_rate': 1 - unique / total if total > 0 else 0,
            'max_cluster': max_c,
            'max_ratio': max_c / total if total > 0 else 0,
            'top3_cluster_sum': top3_sum,
            'cluster_sizes': cluster_sizes[:10]  # 保留前10个簇大小
        }
    
    # =========================================================================
    # 特征增强
    # =========================================================================
    
    def enhance_metrics(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        对分析结果进行特征增强
        
        新增特征：
        - abs_redundancy: 绝对冗余量
        - weighted_rep_score: 流量加权重复分
        - burst_index: 爆发指数
        - sustained_cluster: 持续聚类标记
        - diversity_drop: 多样性跌幅
        - is_valid: 是否超过最小阈值
        - event_score: 综合事件得分
        """
        if df.empty:
            return df
        
        df = df.copy()
        
        # ===== A. 基础门槛 =====
        # 只有图片量超过阈值才是有效数据
        df['is_valid'] = df['img_count'] >= self.min_volume_threshold
        
        # ===== B. 强度特征 =====
        # 绝对冗余量 = 总图片 - 独立组数
        df['abs_redundancy'] = df['img_count'] - df['unique_groups']
        
        # 流量加权重复分 (log抑制超大流量)
        df['weighted_rep_score'] = np.log1p(df['img_count']) * df['rep_rate'].fillna(0)
        
        # ===== C. 聚类特征 =====
        # 爆发指数: 当前max_cluster / 滚动均值
        df['max_cluster_ma'] = df['max_cluster'].rolling(window=4, min_periods=1).mean()
        df['burst_index'] = df['max_cluster'] / (df['max_cluster_ma'] + 0.1)
        
        # 持续聚类检测: 连续N个窗口max_cluster > 阈值
        cluster_threshold = 3
        df['cluster_above'] = (df['max_cluster'] >= cluster_threshold).astype(int)
        df['sustained_cluster'] = df['cluster_above'].rolling(window=3, min_periods=1).sum()
        # sustained_cluster >= 3 表示连续3个窗口都有聚类
        
        # ===== D. 异常检测特征 =====
        # 多样性跌幅: 发帖量上升但独立比例下降
        df['volume_diff'] = df['total_volume'].diff().fillna(0)
        df['rep_rate_diff'] = df['rep_rate'].diff().fillna(0)
        
        # 当发帖量上升且重复率也上升时，标记为异常
        df['diversity_drop'] = ((df['volume_diff'] > 0) & (df['rep_rate_diff'] > 0.1)).astype(int)
        
        # ===== E. 综合事件得分 =====
        df['event_score'] = self._calculate_event_score(df)
        
        # 清理中间列
        df.drop(columns=['cluster_above', 'max_cluster_ma'], inplace=True, errors='ignore')
        
        return df
    
    def _calculate_event_score(self, df: pd.DataFrame) -> pd.Series:
        """
        计算综合事件得分 (0-100)
        
        考虑因素：
        1. 图片量权重 (20%)
        2. 重复率权重 (25%)  
        3. 最大簇权重 (25%)
        4. 持续性权重 (15%)
        5. 异常指标权重 (15%)
        """
        scores = pd.Series(0.0, index=df.index)
        
        # 只对有效数据计算
        valid = df['is_valid']
        
        if not valid.any():
            return scores
        
        # 1. 图片量得分 (归一化到0-20)
        img_max = df.loc[valid, 'img_count'].max()
        if img_max > 0:
            scores.loc[valid] += (df.loc[valid, 'img_count'] / img_max) * 20
        
        # 2. 重复率得分 (0-25)
        scores.loc[valid] += df.loc[valid, 'rep_rate'].fillna(0) * 25
        
        # 3. 最大簇得分 (归一化到0-25)
        cluster_max = df.loc[valid, 'max_cluster'].max()
        if cluster_max > 0:
            scores.loc[valid] += (df.loc[valid, 'max_cluster'] / cluster_max) * 25
        
        # 4. 持续性得分 (0-15)
        scores.loc[valid] += (df.loc[valid, 'sustained_cluster'] / 3).clip(0, 1) * 15
        
        # 5. 异常指标得分 (0-15)
        scores.loc[valid] += df.loc[valid, 'burst_index'].clip(0, 3) / 3 * 10
        scores.loc[valid] += df.loc[valid, 'diversity_drop'] * 5
        
        return scores.round(2)
    
    def detect_events(self, df: pd.DataFrame, score_threshold: float = 50) -> pd.DataFrame:
        """
        检测疑似事件
        
        Args:
            df: enhance_metrics 处理后的 DataFrame
            score_threshold: 事件得分阈值
            
        Returns:
            疑似事件的 DataFrame
        """
        if df.empty:
            return pd.DataFrame()
        
        # 筛选条件
        events = df[
            (df['is_valid']) &  # 超过最小阈值
            (df['event_score'] >= score_threshold)  # 事件得分超过阈值
        ].copy()
        
        if events.empty:
            print("✅ 未检测到明显异常事件")
        else:
            print(f"⚠️ 检测到 {len(events)} 个疑似事件窗口")
        
        return events
    
    # =========================================================================
    # 时序分析
    # =========================================================================
    
    def analyze(self, df: pd.DataFrame, start_date: str, end_date: str,
                image_col: str = 'image_urls', video_col: str = 'video_cover_url',
                time_col: str = 'timestamp', freq: str = '15T') -> pd.DataFrame:
        
        print(f"🔍 分析: {start_date} ~ {end_date}")
        
        df = df.copy()
        df['_dt'] = pd.to_datetime(df[time_col], errors='coerce')
        df = df[(df['_dt'] >= start_date) & (df['_dt'] <= end_date)]
        
        if df.empty:
            print("⚠️ 无数据")
            return pd.DataFrame()
        
        print(f"   数据量: {len(df)} 条")
        
        # 解析路径
        print("📁 解析图片路径...")
        
        def get_paths(row):
            paths = []
            if image_col in row.index:
                for url in self.parse_image_field(row[image_col]):
                    p = self.url_to_local_path(url)
                    if p: paths.append(p)
            if video_col in row.index:
                p = self.url_to_local_path(row[video_col])
                if p: paths.append(p)
            return paths
        
        df['_paths'] = df.apply(get_paths, axis=1)
        df['_has_img'] = df['_paths'].apply(lambda x: 1 if x else 0)
        
        total_found = sum(len(p) for p in df['_paths'])
        print(f"   找到 {total_found} 张本地图片")
        
        if total_found == 0:
            print("⚠️ 未找到图片")
            return pd.DataFrame()
        
        # 预计算哈希
        print("🔢 计算pHash...")
        all_paths = set(p for paths in df['_paths'] for p in paths)
        new_paths = [p for p in all_paths if p not in self.hash_cache]
        
        if new_paths:
            for p in tqdm(new_paths, desc="计算哈希"):
                self.compute_phash(p)
            self.save_cache()
        
        # 按窗口分析
        print(f"📊 按 {freq} 窗口统计...")
        df.set_index('_dt', inplace=True)
        results = []
        
        for t, g in tqdm(list(df.groupby(pd.Grouper(freq=freq))), desc="分析"):
            if g.empty: continue
            
            paths = [p for ps in g['_paths'] for p in ps]
            hashes = {p: self.hash_cache[p] for p in paths if p in self.hash_cache}
            m = self.calc_metrics(hashes)
            
            results.append({
                'time': t,
                'total_volume': len(g),
                'with_img_count': int(g['_has_img'].sum()),
                'with_img_ratio': g['_has_img'].mean(),
                'img_count': m['total'],
                'unique_groups': m['unique'],
                'rep_rate': m['rep_rate'],
                'max_cluster': m['max_cluster'],
                'max_ratio': m['max_ratio'],
                'top3_cluster_sum': m['top3_cluster_sum']
            })
        
        result_df = pd.DataFrame(results)
        
        # 特征增强
        print("🔧 特征增强...")
        result_df = self.enhance_metrics(result_df)
        
        self._summary(result_df)
        return result_df
    
    def _summary(self, df):
        if df.empty: return
        
        valid_df = df[df['is_valid']]
        
        print(f"\n{'='*60}")
        print(f"📋 总体统计")
        print(f"   时间窗口: {len(df)} 个 (有效: {len(valid_df)} 个)")
        print(f"   总帖子: {df['total_volume'].sum()} | 含图: {df['with_img_count'].sum()}")
        
        if len(valid_df) > 0:
            print(f"\n📊 有效窗口统计 (img_count >= {self.min_volume_threshold})")
            print(f"   平均重复率: {valid_df['rep_rate'].mean():.1%}")
            print(f"   最高重复率: {valid_df['rep_rate'].max():.1%}")
            print(f"   最大簇: {valid_df['max_cluster'].max()}")
            print(f"   最大绝对冗余: {valid_df['abs_redundancy'].max()}")
            print(f"   最高事件得分: {valid_df['event_score'].max():.1f}")
            
            # 高风险窗口
            high_risk = valid_df[valid_df['event_score'] >= 50]
            if len(high_risk) > 0:
                print(f"\n⚠️ 高风险窗口 ({len(high_risk)} 个):")
                cols = ['time', 'img_count', 'rep_rate', 'max_cluster', 'abs_redundancy', 'event_score']
                print(high_risk[cols].to_string(index=False))
        
        print('='*60)


# =============================================================================
# 可视化 - 增强版
# =============================================================================

def plot_visual_analysis_enhanced(df: pd.DataFrame, save_path: str = None):
    """增强版可视化 - 6个子图"""
    if df.empty: 
        return
    
    fig, axes = plt.subplots(3, 2, figsize=(16, 12))
    t = df['time']
    valid = df['is_valid']
    
    # ① 总量 + 重复率 + 事件得分
    ax = axes[0, 0]
    ax.bar(t, df['total_volume'], width=0.008, color='#1f77b4', alpha=0.3, label='发帖量')
    ax.set_ylabel('发帖量', color='#1f77b4')
    ax2 = ax.twinx()
    # 重复率 (只画有效窗口)
    rep_rate_plot = df['rep_rate'].where(valid)
    ax2.plot(t, rep_rate_plot, 'r-o', markersize=3, lw=1.5, label='重复率')
    ax2.axhline(0.3, color='orange', ls='--', alpha=0.5)
    ax2.axhline(0.5, color='red', ls='--', alpha=0.5)
    ax2.set_ylabel('重复率', color='red')
    ax2.set_ylim(0, 1.05)
    ax.set_title('① 发帖量 & 重复率 (仅有效窗口)', fontweight='bold')
    
    # ② 含图量 + 含图比例
    ax = axes[0, 1]
    ax.bar(t, df['with_img_count'], width=0.008, color='#2ca02c', alpha=0.5)
    ax.set_ylabel('含图数', color='#2ca02c')
    ax2 = ax.twinx()
    ax2.plot(t, df['with_img_ratio']*100, 'purple', marker='s', markersize=2, lw=1)
    ax2.set_ylabel('含图比例%', color='purple')
    ax2.set_ylim(0, 105)
    ax.set_title('② 含图量 & 含图比例', fontweight='bold')
    
    # ③ 绝对冗余量 (关键新指标)
    ax = axes[1, 0]
    colors = ['#d62728' if v else '#aaaaaa' for v in valid]
    ax.bar(t, df['abs_redundancy'], width=0.008, color=colors, alpha=0.7)
    ax.axhline(df.loc[valid, 'abs_redundancy'].mean() if valid.any() else 0, 
               color='blue', ls='--', label='有效均值')
    ax.set_ylabel('绝对冗余量')
    ax.legend(loc='upper right', fontsize=8)
    ax.set_title('③ 绝对冗余量 (红=有效窗口)', fontweight='bold')
    
    # ④ 最大簇 + Top3簇总和
    ax = axes[1, 1]
    ax.bar(t, df['top3_cluster_sum'], width=0.008, color='#9467bd', alpha=0.4, label='Top3簇总和')
    ax.bar(t, df['max_cluster'], width=0.008, color='#d62728', alpha=0.7, label='最大簇')
    ax.axhline(5, color='orange', ls='--', alpha=0.7)
    ax.set_ylabel('簇大小')
    ax.legend(loc='upper right', fontsize=8)
    ax.set_title('④ 聚类分析 (最大簇 & Top3总和)', fontweight='bold')
    
    # ⑤ 爆发指数 + 持续聚类
    ax = axes[2, 0]
    ax.fill_between(t, 0, df['burst_index'].clip(0, 5), alpha=0.4, color='#ff7f0e', label='爆发指数')
    ax.axhline(2, color='red', ls='--', alpha=0.7, label='爆发阈值')
    ax.set_ylabel('爆发指数', color='#ff7f0e')
    ax2 = ax.twinx()
    ax2.plot(t, df['sustained_cluster'], 'g-', lw=2, label='持续聚类')
    ax2.axhline(3, color='green', ls=':', alpha=0.7)
    ax2.set_ylabel('持续聚类窗口数', color='green')
    ax2.set_ylim(0, 5)
    ax.legend(loc='upper left', fontsize=8)
    ax2.legend(loc='upper right', fontsize=8)
    ax.set_title('⑤ 爆发指数 & 持续聚类', fontweight='bold')
    
    # ⑥ 事件得分 (综合指标)
    ax = axes[2, 1]
    colors = ['#d62728' if s >= 50 else '#2ca02c' if s >= 30 else '#1f77b4' 
              for s in df['event_score']]
    ax.bar(t, df['event_score'], width=0.008, color=colors, alpha=0.7)
    ax.axhline(50, color='red', ls='--', lw=2, label='高风险线 (50)')
    ax.axhline(30, color='orange', ls='--', lw=1.5, label='关注线 (30)')
    ax.set_ylabel('事件得分')
    ax.set_ylim(0, 105)
    ax.legend(loc='upper right', fontsize=8)
    ax.set_title('⑥ 综合事件得分 (红=高风险)', fontweight='bold')
    
    # 格式化
    for ax in axes.flat:
        ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
        plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right', fontsize=8)
    
    plt.suptitle('视觉同质性检测报告 (增强版)', fontsize=14, fontweight='bold')
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"📊 已保存: {save_path}")
    plt.show()


def plot_event_details(df: pd.DataFrame, events: pd.DataFrame, save_path: str = None):
    """绘制事件详情图"""
    if events.empty:
        print("无事件可绘制")
        return
    
    fig, ax = plt.subplots(figsize=(14, 6))
    t = df['time']
    
    # 背景：总发帖量
    ax.fill_between(t, 0, df['total_volume'], alpha=0.2, color='gray', label='发帖量')
    
    # 标记事件窗口
    for _, event in events.iterrows():
        ax.axvline(event['time'], color='red', alpha=0.5, lw=2)
        ax.annotate(f"Score:{event['event_score']:.0f}\n簇:{event['max_cluster']}",
                   xy=(event['time'], event['total_volume']),
                   xytext=(0, 20), textcoords='offset points',
                   fontsize=8, ha='center',
                   bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))
    
    # 事件得分曲线
    ax2 = ax.twinx()
    ax2.plot(t, df['event_score'], 'r-', lw=2, label='事件得分')
    ax2.axhline(50, color='red', ls='--', alpha=0.5)
    ax2.set_ylabel('事件得分', color='red')
    
    ax.set_xlabel('时间')
    ax.set_ylabel('发帖量')
    ax.set_title('事件检测结果', fontsize=14, fontweight='bold')
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right')
    
    ax.legend(loc='upper left')
    ax2.legend(loc='upper right')
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()


# =============================================================================
# 主入口
# =============================================================================

def verify_visual_impact(df, start_date, end_date, 
                         image_base_dir=IMAGE_PATH,
                         freq='15T', 
                         hamming_threshold=10,
                         min_volume_threshold=20,
                         event_score_threshold=50,
                         save_plot=None):
    """
    主入口函数
    
    Args:
        df: 数据框
        start_date, end_date: 时间范围
        image_base_dir: 图片目录
        freq: 时间窗口
        hamming_threshold: pHash相似阈值
        min_volume_threshold: 最小图片量阈值 (过滤噪声)
        event_score_threshold: 事件检测阈值
        save_plot: 保存路径
        
    Returns:
        (result_df, events_df) 分析结果和检测到的事件
    """
    print("="*60)
    print("🔍 视觉同质性检测 (增强版)")
    print("="*60)
    
    analyzer = VisualRepetitionAnalyzer(
        image_base_dir=image_base_dir,
        hamming_threshold=hamming_threshold,
        min_volume_threshold=min_volume_threshold
    )
    
    result = analyzer.analyze(df, start_date, end_date, freq=freq)
    
    if result.empty:
        return result, pd.DataFrame()
    
    # 检测事件
    events = analyzer.detect_events(result, score_threshold=event_score_threshold)
    
    # 可视化
    plot_visual_analysis_enhanced(result, save_path=save_plot)
    
    if not events.empty:
        plot_event_details(result, events, 
                          save_path=save_plot.replace('.png', '_events.png') if save_plot else None)
    
    return result, events

```

## 图片重复率V4

```python
"""
视觉冲击检测 (Visual Impact Detection)
专注于：Bug反馈、玩家抗议、水军刷屏等舆情场景
核心理念：从"比例"转向"绝对量"，检测"量级"和"统治力"
"""

import os
import pickle
import hashlib
from pathlib import Path
from collections import defaultdict

import numpy as np
import pandas as pd
from PIL import Image
import imagehash
from tqdm import tqdm
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 全局配置
# IMAGE_PATH = "./weibo_images"


class VisualImpactDetector:
    """视觉冲击检测器"""
    
    def __init__(self, 
                 image_base_dir: str = IMAGE_PATH,
                 cache_path: str = "./phash_cache.pkl",
                 hamming_threshold: int = 10,
                 min_img_threshold: int = 15):  # 最小图片门槛
        self.image_base_dir = Path(image_base_dir)
        self.cache_path = cache_path
        self.hamming_threshold = hamming_threshold
        self.min_img_threshold = min_img_threshold
        self.hash_cache = self._load_cache()
    
    def _load_cache(self) -> dict:
        if os.path.exists(self.cache_path):
            try:
                with open(self.cache_path, 'rb') as f:
                    return pickle.load(f)
            except:
                pass
        return {}
    
    def save_cache(self):
        with open(self.cache_path, 'wb') as f:
            pickle.dump(self.hash_cache, f)
    
    # =========================================================================
    # URL解析
    # =========================================================================
    
    def parse_urls(self, field) -> list:
        if pd.isna(field) or field == '':
            return []
        return str(field).split(', ')
    
    def url_to_path(self, url: str) -> str | None:
        if not url or pd.isna(url):
            return None
        url = str(url).strip()
        if not url:
            return None
        filename = hashlib.md5(url.encode('utf-8')).hexdigest() + '.jpg'
        path = self.image_base_dir / filename[:2] / filename
        return str(path) if path.exists() else None
    
    # =========================================================================
    # 哈希与聚类
    # =========================================================================
    
    def compute_phash(self, path: str) -> str | None:
        if path in self.hash_cache:
            return self.hash_cache[path]
        try:
            img = Image.open(path).convert('RGB')
            h = str(imagehash.phash(img, hash_size=16))
            self.hash_cache[path] = h
            return h
        except:
            return None
    
    def cluster_images(self, hash_dict: dict) -> list:
        """Union-Find聚类，返回按大小降序的簇列表"""
        if not hash_dict:
            return []
        
        paths = list(hash_dict.keys())
        n = len(paths)
        parent = list(range(n))
        
        def find(x):
            if parent[x] != x:
                parent[x] = find(parent[x])
            return parent[x]
        
        def union(x, y):
            px, py = find(x), find(y)
            if px != py:
                parent[py] = px
        
        hashes = [hash_dict[p] for p in paths]
        for i in range(n):
            for j in range(i + 1, n):
                try:
                    if imagehash.hex_to_hash(hashes[i]) - imagehash.hex_to_hash(hashes[j]) <= self.hamming_threshold:
                        union(i, j)
                except:
                    pass
        
        clusters = defaultdict(list)
        for i in range(n):
            clusters[find(i)].append(paths[i])
        
        return sorted(clusters.values(), key=len, reverse=True)
    
    # =========================================================================
    # 核心指标计算
    # =========================================================================
    
    def calc_window_metrics(self, hash_dict: dict) -> dict:
        """计算单个时间窗口的所有指标"""
        total = len(hash_dict)
        
        if total == 0:
            return {
                'img_count': 0, 'unique_groups': 0,
                'abs_redundancy': 0, 'max_cluster': 0,
                'top3_sum': 0, 'top3_dominance': 0,
                'cluster_sizes': []
            }
        
        clusters = self.cluster_images(hash_dict)
        unique = len(clusters)
        sizes = [len(c) for c in clusters]
        
        max_cluster = sizes[0] if sizes else 0
        top3_sum = sum(sizes[:3])
        
        return {
            'img_count': total,
            'unique_groups': unique,
            'abs_redundancy': total - unique,           # 绝对冗余量
            'max_cluster': max_cluster,                 # 最大簇
            'top3_sum': top3_sum,                       # Top3簇总和
            'top3_dominance': top3_sum / total,         # Top3统治力
            'cluster_sizes': sizes[:10]
        }
    
    def calculate_impact_score(self, row) -> float:
        """
        计算视觉冲击分 (Visual Impact Score)
        
        公式设计：
        1. 门槛过滤：img_count < threshold → 0分
        2. 量级得分：abs_redundancy / 10
        3. 统治力因子：1 + (max_cluster / img_count)
        4. 最终得分 = 量级 × 统治力，映射到0-100
        """
        # 门槛过滤 (过滤低流量噪音)
        if row['img_count'] < self.min_img_threshold:
            return 0.0
        
        # 量级得分：每10张冗余图 = 1分
        magnitude = row['abs_redundancy'] / 10.0
        
        # 统治力因子：最大簇占比越高，系数越大
        dominance = 1 + (row['max_cluster'] / row['img_count'])
        
        # 综合得分
        score = magnitude * dominance
        
        # 映射到 0-100
        return min(100, score * 10)
    
    def enhance_metrics(self, df: pd.DataFrame) -> pd.DataFrame:
        """增强指标计算"""
        if df.empty:
            return df
        
        df = df.copy()
        
        # ===== 1. 基础判定 =====
        df['is_valid'] = df['img_count'] >= self.min_img_threshold
        
        # ===== 2. 流量加权重复分 =====
        # sqrt(img_count) × rep_rate²  — 惩罚低重复率，奖励高重复率
        rep_rate = df['abs_redundancy'] / df['img_count'].replace(0, 1)
        df['weighted_score'] = np.sqrt(df['img_count']) * (rep_rate ** 2)
        
        # ===== 3. 视觉爆发指数 =====
        # 当前max_cluster / 过去4窗口均值
        df['cluster_ma'] = df['max_cluster'].rolling(4, min_periods=1).mean().shift(1).fillna(1)
        df['burst_index'] = df['max_cluster'] / (df['cluster_ma'] + 0.1)
        
        # ===== 4. 持续冲击检测 =====
        # 连续3窗口 max_cluster >= 3
        df['cluster_flag'] = (df['max_cluster'] >= 3).astype(int)
        df['sustained_impact'] = df['cluster_flag'].rolling(3, min_periods=1).sum()
        
        # ===== 5. 视觉冲击分 =====
        df['impact_score'] = df.apply(self.calculate_impact_score, axis=1)
        
        # 清理中间列
        df.drop(columns=['cluster_ma', 'cluster_flag'], inplace=True, errors='ignore')
        
        return df
    
    # =========================================================================
    # 主分析流程
    # =========================================================================
    
    def analyze(self, df: pd.DataFrame, start_date: str, end_date: str,
                image_col: str = 'image_urls', video_col: str = 'video_cover_url',
                time_col: str = 'timestamp', freq: str = '15T') -> pd.DataFrame:
        
        print(f"🔍 视觉冲击检测: {start_date} ~ {end_date}")
        
        df = df.copy()
        df['_dt'] = pd.to_datetime(df[time_col], errors='coerce')
        df = df[(df['_dt'] >= start_date) & (df['_dt'] <= end_date)]
        
        if df.empty:
            print("⚠️ 无数据")
            return pd.DataFrame()
        
        # 解析图片路径
        print("📁 解析图片...")
        def get_paths(row):
            paths = []
            for url in self.parse_urls(row.get(image_col, '')):
                p = self.url_to_path(url)
                if p: paths.append(p)
            if video_col in row:
                p = self.url_to_path(row[video_col])
                if p: paths.append(p)
            return paths
        
        df['_paths'] = df.apply(get_paths, axis=1)
        df['_has_img'] = df['_paths'].apply(bool).astype(int)
        
        total_found = sum(len(p) for p in df['_paths'])
        print(f"   找到 {total_found} 张图片")
        
        if total_found == 0:
            return pd.DataFrame()
        
        # 计算哈希
        print("🔢 计算pHash...")
        all_paths = set(p for ps in df['_paths'] for p in ps)
        new_paths = [p for p in all_paths if p not in self.hash_cache]
        if new_paths:
            for p in tqdm(new_paths, desc="哈希"):
                self.compute_phash(p)
            self.save_cache()
        
        # 按窗口统计
        print(f"📊 按 {freq} 窗口分析...")
        df.set_index('_dt', inplace=True)
        results = []
        
        for t, g in tqdm(list(df.groupby(pd.Grouper(freq=freq))), desc="分析"):
            if g.empty: continue
            
            paths = [p for ps in g['_paths'] for p in ps]
            hashes = {p: self.hash_cache[p] for p in paths if p in self.hash_cache}
            m = self.calc_window_metrics(hashes)
            
            results.append({
                'time': t,
                'total_volume': len(g),
                'with_img_count': int(g['_has_img'].sum()),
                'with_img_ratio': g['_has_img'].mean(),
                **m
            })
        
        result_df = pd.DataFrame(results)
        result_df = self.enhance_metrics(result_df)
        
        self._print_summary(result_df)
        return result_df
    
    def _print_summary(self, df):
        if df.empty: return
        
        valid = df[df['is_valid']]
        
        print(f"\n{'='*60}")
        print(f"📋 分析摘要 (门槛: {self.min_img_threshold} 张图)")
        print(f"   窗口数: {len(df)} (有效: {len(valid)})")
        print(f"   总帖子: {df['total_volume'].sum()} | 含图: {df['with_img_count'].sum()}")
        
        if len(valid) > 0:
            print(f"\n📊 有效窗口指标:")
            print(f"   最大绝对冗余: {valid['abs_redundancy'].max()}")
            print(f"   最大簇: {valid['max_cluster'].max()}")
            print(f"   最高Top3统治力: {valid['top3_dominance'].max():.1%}")
            print(f"   最高冲击分: {valid['impact_score'].max():.1f}")
            
            # 高危窗口
            high_impact = valid[valid['impact_score'] >= 30]
            if len(high_impact) > 0:
                print(f"\n⚠️ 高冲击窗口 ({len(high_impact)}个):")
                cols = ['time', 'img_count', 'abs_redundancy', 'max_cluster', 
                        'top3_dominance', 'impact_score']
                print(high_impact[cols].to_string(index=False))
        
        print('='*60)
    
    def detect_events(self, df: pd.DataFrame, 
                      impact_threshold: float = 30,
                      burst_threshold: float = 3) -> pd.DataFrame:
        """
        检测舆情事件
        
        条件 (满足任一):
        1. impact_score >= impact_threshold
        2. burst_index >= burst_threshold 且 is_valid
        3. sustained_impact >= 3 (连续3窗口有聚类)
        """
        if df.empty:
            return pd.DataFrame()
        
        events = df[
            (df['impact_score'] >= impact_threshold) |
            ((df['burst_index'] >= burst_threshold) & df['is_valid']) |
            (df['sustained_impact'] >= 3)
        ].copy()
        
        if events.empty:
            print("✅ 未检测到舆情冲击事件")
        else:
            print(f"🚨 检测到 {len(events)} 个疑似冲击事件")
        
        return events


# =============================================================================
# 可视化
# =============================================================================

def plot_impact_analysis(df: pd.DataFrame, min_threshold: int = 15, save_path: str = None):
    """舆情冲击可视化 - 6图布局"""
    if df.empty: return
    
    fig, axes = plt.subplots(3, 2, figsize=(16, 12))
    t = df['time']
    valid = df['is_valid']
    
    # ① 绝对冗余量 (核心指标) + 发帖量背景
    ax = axes[0, 0]
    ax.fill_between(t, 0, df['total_volume'], alpha=0.15, color='gray', label='发帖量')
    colors = ['#d62728' if v else '#cccccc' for v in valid]
    ax.bar(t, df['abs_redundancy'], width=0.008, color=colors, alpha=0.8, label='绝对冗余')
    ax.axhline(10, color='orange', ls='--', alpha=0.7, label='关注线(10)')
    ax.set_ylabel('绝对冗余量')
    ax.legend(loc='upper right', fontsize=8)
    ax.set_title('① 绝对冗余量 (红=有效窗口，灰=低流量噪音)', fontweight='bold')
    
    # ② Top3统治力 + 最大簇
    ax = axes[0, 1]
    ax.bar(t, df['top3_sum'], width=0.008, color='#9467bd', alpha=0.4, label='Top3簇总和')
    ax.bar(t, df['max_cluster'], width=0.008, color='#d62728', alpha=0.8, label='最大簇')
    ax2 = ax.twinx()
    dom_plot = df['top3_dominance'].where(valid) * 100
    ax2.plot(t, dom_plot, 'b-o', markersize=2, lw=1, label='Top3统治力%')
    ax2.axhline(40, color='blue', ls='--', alpha=0.5)
    ax2.set_ylabel('统治力%', color='blue')
    ax2.set_ylim(0, 105)
    ax.set_ylabel('簇大小')
    ax.legend(loc='upper left', fontsize=8)
    ax2.legend(loc='upper right', fontsize=8)
    ax.set_title('② 头部统治力 (Top3占比 > 40% = 高风险)', fontweight='bold')
    
    # ③ 视觉冲击分 (综合指标)
    ax = axes[1, 0]
    colors = ['#d62728' if s >= 50 else '#ff7f0e' if s >= 30 else '#2ca02c' 
              for s in df['impact_score']]
    ax.bar(t, df['impact_score'], width=0.008, color=colors, alpha=0.8)
    ax.axhline(50, color='red', ls='--', lw=2, label='高危(50)')
    ax.axhline(30, color='orange', ls='--', lw=1.5, label='警告(30)')
    ax.set_ylabel('冲击分')
    ax.set_ylim(0, max(105, df['impact_score'].max() * 1.1))
    ax.legend(loc='upper right', fontsize=8)
    ax.set_title('③ 视觉冲击分 (量级×统治力)', fontweight='bold')
    
    # ④ 爆发指数
    ax = axes[1, 1]
    ax.fill_between(t, 0, df['burst_index'].clip(0, 10), alpha=0.6, color='#ff7f0e')
    ax.axhline(3, color='red', ls='--', lw=2, label='爆发阈值(3x)')
    ax.axhline(2, color='orange', ls='--', lw=1, label='关注(2x)')
    ax.set_ylabel('爆发指数 (vs 历史均值)')
    ax.set_ylim(0, max(5, df['burst_index'].max() * 1.1))
    ax.legend(loc='upper right', fontsize=8)
    ax.set_title('④ 爆发指数 (当前/历史均值)', fontweight='bold')
    
    # ⑤ 含图量 + 含图比例
    ax = axes[2, 0]
    ax.bar(t, df['with_img_count'], width=0.008, color='#2ca02c', alpha=0.5)
    ax.axhline(min_threshold, color='gray', ls=':', label=f'有效门槛({min_threshold})')
    ax.set_ylabel('含图帖数', color='#2ca02c')
    ax2 = ax.twinx()
    ax2.plot(t, df['with_img_ratio']*100, 'purple', marker='s', markersize=2, lw=1)
    ax2.set_ylabel('含图比例%', color='purple')
    ax2.set_ylim(0, 105)
    ax.legend(loc='upper right', fontsize=8)
    ax.set_title('⑤ 含图量 & 比例', fontweight='bold')
    
    # ⑥ 持续冲击 + 流量加权分
    ax = axes[2, 1]
    ax.bar(t, df['sustained_impact'], width=0.008, color='#17becf', alpha=0.6, label='持续冲击窗口数')
    ax.axhline(3, color='red', ls='--', label='持续事件阈值(3)')
    ax2 = ax.twinx()
    ws_plot = df['weighted_score'].where(valid)
    ax2.plot(t, ws_plot, 'r-', lw=1.5, label='流量加权分')
    ax2.set_ylabel('加权分', color='red')
    ax.set_ylabel('持续窗口数')
    ax.set_ylim(0, 5)
    ax.legend(loc='upper left', fontsize=8)
    ax2.legend(loc='upper right', fontsize=8)
    ax.set_title('⑥ 持续冲击 & 加权得分', fontweight='bold')
    
    # 格式化
    for ax in axes.flat:
        ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
        plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right', fontsize=8)
    
    plt.suptitle('舆情视觉冲击检测报告', fontsize=14, fontweight='bold')
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"📊 已保存: {save_path}")
    plt.show()


def plot_event_timeline(df: pd.DataFrame, events: pd.DataFrame, save_path: str = None):
    """事件时间线"""
    if events.empty:
        return
    
    fig, ax = plt.subplots(figsize=(14, 5))
    t = df['time']
    
    # 背景
    ax.fill_between(t, 0, df['total_volume'], alpha=0.2, color='gray')
    ax.bar(t, df['abs_redundancy'], width=0.008, alpha=0.5, color='#d62728')
    
    # 标记事件
    for _, e in events.iterrows():
        ax.axvline(e['time'], color='red', alpha=0.7, lw=2)
        label = f"冲击:{e['impact_score']:.0f}\n冗余:{e['abs_redundancy']}\n簇:{e['max_cluster']}"
        ax.annotate(label, xy=(e['time'], e['abs_redundancy']),
                   xytext=(5, 30), textcoords='offset points', fontsize=7,
                   bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.8),
                   arrowprops=dict(arrowstyle='->', color='red'))
    
    ax.set_ylabel('绝对冗余量')
    ax.set_title(f'舆情冲击事件时间线 (检测到 {len(events)} 个事件)', fontweight='bold')
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right')
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()


# =============================================================================
# 主入口
# =============================================================================

def verify_visual_impact(df, start_date, end_date, 
                         image_base_dir=IMAGE_PATH,
                         freq='15T',
                         hamming_threshold=10,
                         min_img_threshold=15,
                         impact_threshold=30,
                         save_plot=None):
    """
    视觉冲击检测主入口
    
    Args:
        df: 数据框
        start_date, end_date: 时间范围
        image_base_dir: 图片目录
        freq: 时间窗口 (如 '15T', '30T', '1H')
        hamming_threshold: pHash相似阈值 (越小越严格)
        min_img_threshold: 最小图片量门槛 (过滤低流量噪音)
        impact_threshold: 冲击分报警阈值
        save_plot: 保存路径
        
    Returns:
        (result_df, events_df)
    """
    print("="*60)
    print("🚨 舆情视觉冲击检测")
    print(f"   门槛: {min_img_threshold}张图 | 报警: {impact_threshold}分")
    print("="*60)
    
    detector = VisualImpactDetector(
        image_base_dir=image_base_dir,
        hamming_threshold=hamming_threshold,
        min_img_threshold=min_img_threshold
    )
    
    result = detector.analyze(df, start_date, end_date, freq=freq)
    
    if result.empty:
        return result, pd.DataFrame()
    
    events = detector.detect_events(result, impact_threshold=impact_threshold)
    
    # 可视化
    plot_impact_analysis(result, min_img_threshold, save_path=save_plot)
    
    if not events.empty and save_plot:
        plot_event_timeline(result, events, 
                           save_path=save_plot.replace('.png', '_events.png'))
    
    return result, events

```

### final

```python
"""
视觉冲击检测 - 精简版
核心特征：
1. vis_abs_redundancy: 视觉绝对冗余量 = 含图总数 - 视觉独立组数
2. vis_concentration: 视觉集中度 = Top3_Sum / (Total + K), K=15
"""

import os
import pickle
import hashlib
from pathlib import Path
from collections import defaultdict

import numpy as np
import pandas as pd
from PIL import Image
import imagehash
from tqdm import tqdm
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# IMAGE_PATH = "./weibo_images"
DAMPING_K = 15  # 阻尼系数


class VisualImpactDetector:
    """视觉冲击检测器 - 精简版"""
    
    def __init__(self, image_base_dir: str = IMAGE_PATH,
                 cache_path: str = "./phash_cache.pkl",
                 hamming_threshold: int = 10):
        self.image_base_dir = Path(image_base_dir)
        self.cache_path = cache_path
        self.hamming_threshold = hamming_threshold
        self.hash_cache = self._load_cache()
    
    def _load_cache(self) -> dict:
        if os.path.exists(self.cache_path):
            try:
                with open(self.cache_path, 'rb') as f:
                    return pickle.load(f)
            except:
                pass
        return {}
    
    def save_cache(self):
        with open(self.cache_path, 'wb') as f:
            pickle.dump(self.hash_cache, f)
    
    def parse_urls(self, field) -> list:
        if pd.isna(field) or field == '':
            return []
        return str(field).split(', ')
    
    def url_to_path(self, url: str) -> str | None:
        if not url or pd.isna(url):
            return None
        url = str(url).strip()
        if not url:
            return None
        filename = hashlib.md5(url.encode('utf-8')).hexdigest() + '.jpg'
        path = self.image_base_dir / filename[:2] / filename
        return str(path) if path.exists() else None
    
    def compute_phash(self, path: str) -> str | None:
        if path in self.hash_cache:
            return self.hash_cache[path]
        try:
            img = Image.open(path).convert('RGB')
            h = str(imagehash.phash(img, hash_size=16))
            self.hash_cache[path] = h
            return h
        except:
            return None
    
    def cluster_images(self, hash_dict: dict) -> list:
        """Union-Find聚类，返回按大小降序的簇"""
        if not hash_dict:
            return []
        
        paths = list(hash_dict.keys())
        n = len(paths)
        parent = list(range(n))
        
        def find(x):
            if parent[x] != x:
                parent[x] = find(parent[x])
            return parent[x]
        
        def union(x, y):
            px, py = find(x), find(y)
            if px != py:
                parent[py] = px
        
        hashes = [hash_dict[p] for p in paths]
        for i in range(n):
            for j in range(i + 1, n):
                try:
                    if imagehash.hex_to_hash(hashes[i]) - imagehash.hex_to_hash(hashes[j]) <= self.hamming_threshold:
                        union(i, j)
                except:
                    pass
        
        clusters = defaultdict(list)
        for i in range(n):
            clusters[find(i)].append(paths[i])
        
        return sorted(clusters.values(), key=len, reverse=True)
    
    def calc_metrics(self, hash_dict: dict) -> dict:
        """计算两个核心指标"""
        total = len(hash_dict)
        
        if total == 0:
            return {'img_count': 0, 'unique_groups': 0, 'top3_sum': 0,
                    'vis_abs_redundancy': 0, 'vis_concentration': 0}
        
        clusters = self.cluster_images(hash_dict)
        unique = len(clusters)
        sizes = [len(c) for c in clusters]
        top3_sum = sum(sizes[:3])
        
        return {
            'img_count': total,
            'unique_groups': unique,
            'top3_sum': top3_sum,
            # 核心特征1: 绝对冗余量
            'vis_abs_redundancy': total - unique,
            # 核心特征2: 视觉集中度 (阻尼比例)
            'vis_concentration': top3_sum / (total + DAMPING_K)
        }
    
    def analyze(self, df: pd.DataFrame, start_date: str, end_date: str,
                image_col: str = 'image_urls', video_col: str = 'video_cover_url',
                time_col: str = 'timestamp', freq: str = '15T') -> pd.DataFrame:
        
        print(f"🔍 分析: {start_date} ~ {end_date}")
        
        df = df.copy()
        df['_dt'] = pd.to_datetime(df[time_col], errors='coerce')
        df = df[(df['_dt'] >= start_date) & (df['_dt'] <= end_date)]
        
        if df.empty:
            print("⚠️ 无数据")
            return pd.DataFrame()
        
        # 解析路径
        print("📁 解析图片...")
        def get_paths(row):
            paths = []
            for url in self.parse_urls(row.get(image_col, '')):
                p = self.url_to_path(url)
                if p: paths.append(p)
            if video_col in row:
                p = self.url_to_path(row[video_col])
                if p: paths.append(p)
            return paths
        
        df['_paths'] = df.apply(get_paths, axis=1)
        df['_has_img'] = df['_paths'].apply(bool).astype(int)
        
        total_found = sum(len(p) for p in df['_paths'])
        print(f"   找到 {total_found} 张图片")
        
        if total_found == 0:
            return pd.DataFrame()
        
        # 计算哈希
        print("🔢 计算pHash...")
        all_paths = set(p for ps in df['_paths'] for p in ps)
        new_paths = [p for p in all_paths if p not in self.hash_cache]
        if new_paths:
            for p in tqdm(new_paths, desc="哈希"):
                self.compute_phash(p)
            self.save_cache()
        
        # 按窗口统计
        print(f"📊 按 {freq} 窗口分析...")
        df.set_index('_dt', inplace=True)
        results = []
        
        for t, g in tqdm(list(df.groupby(pd.Grouper(freq=freq))), desc="分析"):
            if g.empty: continue
            
            paths = [p for ps in g['_paths'] for p in ps]
            hashes = {p: self.hash_cache[p] for p in paths if p in self.hash_cache}
            m = self.calc_metrics(hashes)
            
            results.append({
                'time': t,
                'total_volume': len(g),
                'with_img_count': int(g['_has_img'].sum()),
                **m
            })
        
        result_df = pd.DataFrame(results)
        self._summary(result_df)
        return result_df
    
    def _summary(self, df):
        if df.empty: return
        print(f"\n{'='*50}")
        print(f"📋 窗口: {len(df)} | 帖子: {df['total_volume'].sum()} | 图片: {df['img_count'].sum()}")
        print(f"   最大绝对冗余: {df['vis_abs_redundancy'].max()}")
        print(f"   最高集中度: {df['vis_concentration'].max():.3f}")
        print('='*50)


def plot_visual_impact(df: pd.DataFrame, save_path: str = None):
    """可视化两个核心特征"""
    if df.empty: return
    
    fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)
    t = df['time']
    
    # ① 绝对冗余量 (柱) + 发帖量背景
    ax = axes[0]
    ax.fill_between(t, 0, df['total_volume'], alpha=0.15, color='gray', label='发帖量')
    ax.bar(t, df['vis_abs_redundancy'], width=0.008, color='#d62728', alpha=0.8, label='视觉绝对冗余')
    ax.set_ylabel('绝对冗余量')
    ax.legend(loc='upper right')
    ax.set_title('① vis_abs_redundancy (含图数 - 独立组数)', fontweight='bold')
    
    # ② 视觉集中度 (线) + 含图量背景
    ax = axes[1]
    ax.bar(t, df['img_count'], width=0.008, color='#1f77b4', alpha=0.3, label='含图数')
    ax2 = ax.twinx()
    ax2.plot(t, df['vis_concentration'], 'r-o', markersize=3, lw=1.5, label='视觉集中度')
    ax2.axhline(0.5, color='orange', ls='--', alpha=0.7, label='关注线(0.5)')
    ax2.set_ylabel('集中度', color='red')
    ax2.set_ylim(0, 1.05)
    ax.set_ylabel('含图数')
    ax.legend(loc='upper left')
    ax2.legend(loc='upper right')
    ax.set_title(f'② vis_concentration = Top3 / (Total + {DAMPING_K})', fontweight='bold')
    
    # 格式化
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right')
    
    plt.suptitle('视觉冲击检测 - 核心特征', fontsize=14, fontweight='bold')
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"📊 已保存: {save_path}")
    plt.show()


def verify_visual_impact(df, start_date, end_date,
                         image_base_dir=IMAGE_PATH,
                         freq='15T',
                         hamming_threshold=10,
                         save_plot=None) -> pd.DataFrame:
    """
    主入口
    
    Returns:
        DataFrame with columns:
        - time, total_volume, with_img_count, img_count, unique_groups, top3_sum
        - vis_abs_redundancy: 绝对冗余量
        - vis_concentration: 视觉集中度 (阻尼比例)
    """
    detector = VisualImpactDetector(image_base_dir, hamming_threshold=hamming_threshold)
    result = detector.analyze(df, start_date, end_date, freq=freq)
    
    if not result.empty:
        plot_visual_impact(result, save_path=save_plot)
    
    return result

```
